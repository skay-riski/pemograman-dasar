```python
bilangan1 = 10
bilangan2 = 8

print("Bilangan pertama:", bilangan1)
print("Bilangan kedua:", bilangan2)

print("Hasil penjumlahan (+):", bilangan1 + bilangan2)
print("Hasil pengurangan (-):", bilangan1 - bilangan2)
print("Hasil perkalian (*):", bilangan1 * bilangan2)
print("Hasil pembagian (/):", bilangan1 / bilangan2)
print("Hasil pembagian bulat (//):", bilangan1 // bilangan2)
print("Hasil modulus (%):", bilangan1 % bilangan2)


nilai = 75
lulus = nilai >= 60

print("Nilai ini termasuk kategori lulus (>= 60):", lulus)


umur = 21
punya_sim = True

boleh_mengemudi = (umur >= 17) and punya_sim
cek_syarat = (umur >= 17) and (punya_sim == True)

print("Boleh mengemudi sendiri:", boleh_mengemudi)
print("Hasil pengecekan syarat:", cek_syarat)



energi_input = 1000
energi_output = 850

efisiensi = (energi_output / energi_input) * 100

print(f"Efisiensi termal unit pembangkit adalah {efisiensi:.2f}%")


suhu_boiler = 560
beban_operasi = 90

perlu_maintenance = (suhu_boiler > 550) or (beban_operasi > 95)

print(f"Apakah unit pembangkit perlu maintenance? {perlu_maintenance}")
```
