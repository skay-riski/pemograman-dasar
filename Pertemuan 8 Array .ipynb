```python
suhu = [28, 31, 30, 33, 29]       # suhu (°C) pada 5 hari

print("suhu[0] =", suhu[0])       # elemen pertama
print("suhu[4] =", suhu[4])       # elemen terakhir (len - 1)
print("suhu[-1] =", suhu[-1])     # indeks negatif: dihitung dari belakang

# --- tebak dulu, baru jalankan ---
print(suhu[1])
print(suhu[-2])
print(suhu[len(suhu) - 1])
print(suhu[2] + suhu[3])
```

```python
suhu = [28, 31, 30, 33, 29]  # Data suhu selama 5 hari (°C)

print("Suhu hari pertama =", suhu[0])
print("Suhu hari kelima =", suhu[4])
print("Suhu terakhir =", suhu[-1])

# Membaca elemen menggunakan indeks
print("Suhu indeks 1 =", suhu[1])
print("Suhu indeks -2 =", suhu[-2])
print("Suhu terakhir =", suhu[len(suhu) - 1])
print("Jumlah suhu indeks 2 dan 3 =", suhu[2] + suhu[3])
```

```python
hari = ["Sen", "Sel", "Rab", "Kam", "Jum", "Sab", "Min"]

print(hari[1:4])    # indeks 1, 2, 3 (indeks 4 TIDAK ikut)
print(hari[:3])     # dari awal sampai indeks 2
print(hari[4:])     # dari indeks 4 sampai akhir
print(hari[-2:])    # dua elemen terakhir
print(hari[:])      # semuanya: salinan seluruh list
```

```python
hari = ["Sen", "Sel", "Rab", "Kam", "Jum", "Sab", "Min"]

print("Indeks 1 sampai 3 =", hari[1:4])
print("Dari awal sampai indeks 2 =", hari[:3])
print("Indeks 4 sampai akhir =", hari[4:])
print("Dua hari terakhir =", hari[-2:])
print("Seluruh isi list =", hari[:])
```

```python
belanja = ["nasi", "telur", "susu"]

print("Awal           :", belanja)

belanja[1] = "tahu"
print("Setelah ubah   :", belanja)

belanja.append("teh")
print("Setelah append :", belanja)

belanja.insert(0, "garam")
print("Setelah insert :", belanja)

belanja.remove("nasi")
print("Setelah remove :", belanja)

terakhir = belanja.pop()
print("Di-pop         :", terakhir)
print("Setelah pop    :", belanja)

print("Tahu ada?      :", "tahu" in belanja)
print("Nasi ada?      :", "nasi" in belanja)
```

```python

# Data nilai 3 mahasiswa untuk 3 ujian
nilai = [
    [80, 70, 90],  # Ana
    [65, 75, 60],  # Budi
    [90, 85, 95]   # Cici
]

# Menampilkan data
print("Nilai Budi:", nilai[1])
print("Nilai ujian 3 Budi:", nilai[1][2])
print("Jumlah baris:", len(nilai))
print("Jumlah kolom:", len(nilai[0]))

# Menghitung total nilai setiap mahasiswa
for b in range(len(nilai)):
    print("Mahasiswa", b + 1, "total:", sum(nilai[b]))

# Mengambil nilai ujian pertama semua mahasiswa
ujian1 = []

for baris in nilai:
    ujian1.append(baris[0])

print("Skor ujian 1:", ujian1)
```
