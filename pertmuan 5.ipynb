```python
harga_buku = 18500
jumlah_buku = 7
uang_budi = 150000

total_belanja = harga_buku * jumlah_buku
diskon = total_belanja * 10 // 100
total_bayar = total_belanja - diskon
kembalian = uang_budi - total_bayar

iuran = total_bayar // 4
sisa = total_bayar % 4
cukup = uang_budi >= total_bayar

print("Total belanja:", total_belanja)
print("Diskon:", diskon)
print("Total bayar:", total_bayar)
print("Uang kembalian:", kembalian)
print("Iuran per orang:", iuran)
print("Sisa uang:", sisa)
print("Apakah uang Budi cukup?", cukup)

jumlah_panel = 12
daya_panel = 450
jam_sinar = 5
efisiensi = 0.8
kebutuhan = 25
tarif = 1444.70

energi_harian = jumlah_panel * daya_panel / 1000 * jam_sinar * efisiensi
energi_bulanan = energi_harian * 30
nilai_listrik = energi_bulanan * tarif
kekurangan = max(0, kebutuhan - energi_harian)
cukup = energi_harian >= kebutuhan

print("Energi per hari:", energi_harian, "kWh")
print("Energi per bulan:", energi_bulanan, "kWh")
print("Nilai listrik per bulan: Rp", round(nilai_listrik, 2))
print("Kekurangan energi per hari:", kekurangan, "kWh")
print("Apakah PLTS mencukupi?", cukup)

daya_generator = 200
jam_operasi = 8
konsumsi_solar = 0.25
harga_solar = 13500
kapasitas_tangki = 1000
arus = 20
hambatan = 0.5

energi_harian = daya_generator * jam_operasi
solar_harian = energi_harian * konsumsi_solar
biaya_solar = solar_harian * harga_solar

hari_cukup = kapasitas_tangki // solar_harian
sisa_solar = kapasitas_tangki % solar_harian

rugi_daya = arus ** 2 * hambatan
rugi_tinggi = rugi_daya > 150 and arus >= 20

print("Energi per hari:", energi_harian, "kWh")
print("Kebutuhan solar:", solar_harian, "liter")
print("Biaya solar: Rp", biaya_solar)
print("Tangki cukup:", hari_cukup, "hari")
print("Sisa solar:", sisa_solar, "liter")
print("Rugi daya kabel:", rugi_daya, "W")
print("Apakah rugi kabel tinggi?", rugi_tinggi)
```
