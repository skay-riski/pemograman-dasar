```python
daya_kw = 450
jam = 8
energi_aktual = 7200
kapasitas_kw = 400

energi = daya_kw * jam

capacity_factor = (energi_aktual / (kapasitas_kw * 24)) * 100

print(f"Energi harian = {energi} kWh")
print(f"Capacity factor = {capacity_factor:.2f}%")
```

```python
Program ini digunakan untuk menghitung energi listrik harian dan capacity factor. Hasilnya adalah energi harian sebesar 3.600 kWh dan capacity factor sebesar 75%.
```

```python
suhu = float(input("Masukkan suhu (C): "))
tekanan = float(input("Masukkan tekanan (bar): "))

if suhu >= 550 or tekanan >= 14:
    print("STATUS: TRIP")
elif suhu > 520 and tekanan > 12:
     print("STATUS: WARNING")
else:
    print("STATUS: NORMAL")
```

```python
Program ini digunakan untuk memeriksa suhu dan tekanan. Hasilnya menentukan kondisi sistem, yaitu NORMAL, WARNING, atau TRIP sesuai nilai yang dimasukkan.
```

```python
volume = 100

while volume > 0:
    volume -= 20
    print("Volume:", volume, "liter")

    if volume == 20:
        print("WARNING: Hampir kosong!")

print("TANGKI KOSONG - Pengosongan berhenti.")
```

```python
Program ini menyimulasikan pengosongan tangki dari 100 liter hingga kosong, berkurang 20 liter setiap proses, dan menampilkan peringatan saat tersisa 20 liter
```

```python
data = ["62.5", "58.0", "81.3", "N/A", "66.0"]

jumlah_valid = 0

for i, nilai in enumerate(data, start=1):
    try:
        suhu = float(nilai)
        jumlah_valid += 1


        if suhu > 70.0:
            print(f"Warning: Titik {i} suhu bearing tinggi!")

    except ValueError:
        print(f"Titik {i}: data tidak valid - pemeriksaan dihentikan")
        break

print(f"Jumlah data valid: {jumlah_valid}")
```

```python
Program ini digunakan untuk memeriksa suhu bearing. Data diubah menjadi angka, mendeteksi suhu tinggi, dan menghentikan pemeriksaan jika terdapat data tidak valid.
```
